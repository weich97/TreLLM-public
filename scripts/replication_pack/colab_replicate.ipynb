# TradeArena Deterministic Replication Pack v1 (Colab)

Zero-key deterministic replication of the TradeArena benchmark anchor arms.
No API keys, no third-party packages; total runtime is typically 5-15 minutes.

Steps: run the three cells below in order. Cell 1 uploads (or downloads) the pack zip,
cell 2 runs the replication with the Colab environment class, cell 3 downloads the report files
that you send back.

```python
# Cell 1: get the pack zip into Colab.
# Option A (default): upload the zip you received.
# Option B: if your invitation names a release URL, replace PACK_URL and set USE_URL = True.
import pathlib
import zipfile

USE_URL = False
PACK_URL = ""  # e.g. a GitHub release asset URL from your invitation
if USE_URL and PACK_URL:
    import urllib.request
    urllib.request.urlretrieve(PACK_URL, "tradearena_replication_pack_v1.zip")
    zip_name = "tradearena_replication_pack_v1.zip"
else:
    from google.colab import files
    uploaded = files.upload()
    zip_name = next(iter(uploaded))
with zipfile.ZipFile(zip_name) as archive:
    archive.extractall(".")
pack_root = next(p for p in pathlib.Path(".").iterdir() if p.is_dir() and (p / "replicate.py").exists())
print("Pack root:", pack_root)
```
